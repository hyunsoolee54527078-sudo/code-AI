```
from pathlib import Path

# 1. 2-name2text.txt 파일의 형식을 GPT-SoVITS 학습기가 인식할 수 있도록 파이프(|) 포맷으로 교정합니다.
phoneme_path = Path("/content/GPT-SoVITS/logs/asr_opt/2-name2text.txt")

if phoneme_path.exists():
    lines = phoneme_path.read_text(encoding="utf-8").splitlines()
    fixed_lines = []
    for line in lines:
        if not line.strip():
            continue
        # 기존 파일이 탭(\t)으로 구분되어 있는 경우 파이프(|) 포맷으로 변환
        if "\t" in line:
            parts = line.split("\t")
            name = parts[0]
            phonemes = parts[1]
            # 임시로 원본 텍스트도 음소 문자열을 합친 값 등으로 채워 포맷(이름|음소|텍스트)을 강제로 맞춰줍니다.
            clean_text = phonemes.replace(" ", "").replace("空", " ")
            fixed_line = f"{name}|{phonemes}|{clean_text}"
            fixed_lines.append(fixed_line)
        else:
            fixed_lines.append(line)

    phoneme_path.write_text("\n".join(fixed_lines) + "\n", encoding="utf-8")
    print(f"성공적으로 {len(fixed_lines)}개의 학습 데이터 포맷을 교정했습니다!")
    print("수정된 파일 앞부분 2줄:")
    for l in fixed_lines[:2]:
        print(l[:150])
else:
    print("교정할 2-name2text.txt 파일이 존재하지 않습니다. 경로를 확인해주세요.")
```

```
import shutil
from pathlib import Path

# 디렉토리 충돌 방지를 위해 출력 폴더 구조를 클리어하고 재지정합니다.
input_dir = Path("/content/output")
asr_out_dir = Path("/content/asr_opt_fixed")

# 만약 입력 폴더 안에 출력 폴더가 숨어있다면 제거하거나 비워줍니다.
nested_out = input_dir / "asr_opt"
if nested_out.exists():
    print(f"입력 폴더 내부의 출력 경로 {nested_out}를 정리합니다.")
    shutil.rmtree(nested_out, ignore_errors=True)

# 새로운 격리된 출력 디렉토리 생성
asr_out_dir.mkdir(parents=True, exist_ok=True)
print(f"새로운 독립된 출력 경로를 생성했습니다: {asr_out_dir}")
```
